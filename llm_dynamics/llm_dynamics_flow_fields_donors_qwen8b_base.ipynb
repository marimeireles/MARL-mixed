# The arrow fields — donors game

Companion to `llm_dynamics_notebook.ipynb` (the source of truth for methods,
registry and tables). Everything here is generated by
`python -m llm_dynamics.flow_grids` from the logged games — no LLM calls.

Reading the figures: arrows are the CRLD learning flow $\\Delta X$ in the
cooperation plane (model $P(C)$ on x, partner $P(C)$ on y, in the
all-cooperate memory state), i.e. where a small adaptive learner would move
from each joint policy; arrow colour = flow magnitude (viridis: dark = slow,
yellow = fast). Red dots are where the measured base-Qwen3-32B game *ended*
(sliding-window realized $P(C)$ of both players over the last 8 rounds), one
dot per seed. The LLM does not learn within a game — its end point is where
its fixed policy and the partner's rule carried the state — so a dot sitting
against the arrows is an LLM holding a point the learning dynamics would
leave. Background memory equals the LLM's memory where CRLD is feasible
(≤ 3), else memory 1.
